```python
import numpy as np
vetor1 = np.array([1, 2, 3])
vetor2 = np.array([4, 5, 6])
print(vetor1, vetor2)
```

```python
soma = vetor1 + vetor2
print("Soma:", soma)
```

```python
from numpy.linalg import norm
distancia = norm(vetor1 - vetor2)
print("Distância Euclidiana:", distancia)
```

```python
similaridade = np.dot(vetor1, vetor2) / (norm(vetor1) * norm(vetor2))
print("Similaridade Cosseno:", similaridade)
```

```python
import matplotlib.pyplot as plt
vetor_a = np.array([2, 3])
vetor_b = np.array([4, 1])
#vetor_c = np.array([2, 4])
#vetor_c = np.array([2, 3])
vetor_c = np.array([3, 5])
```

```python
plt.quiver(0,0, vetor_a[0], vetor_a[1], angles='xy', scale_units='xy', scale=1, color='r')
plt.quiver(0,0, vetor_b[0], vetor_b[1], angles='xy', scale_units='xy', scale=1, color='b')
plt.quiver(0,0, vetor_c[0], vetor_c[1], angles='xy', scale_units='xy', scale=1, color='g')
plt.xlim(-1, 5)
plt.ylim(-1, 5)
plt.grid()
plt.show()
```

```python
similaridade = np.dot(vetor_c, vetor_a) / (norm(vetor_c) * norm(vetor_a))
print("Similaridade Cosseno:", similaridade)
```
